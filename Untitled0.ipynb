```python
a = list(map(int, input().split()))
for i in range(0, len(a) - 1, 2): a[i], a[i + 1] = a[i + 1], a[i]
print(*a)
```

```python
a = list(map(int, input().split()))

a.insert(0, a.pop())

print(*a)
```

```python
a = input().split()

for i in range(len(a)):
    if a.count(a[i]) == 1:
        print(a[i], end=' ')
```

```python
a = input().split()
best = a[0]

for x in a:
    if a.count(x) > a.count(best):
        best = x

print(best)
```

```python
data = input().split()
n = int(data[0])
a = data[1:]
mid = n // 2

for x in a:
    less = 0
    for y in a:
        if int(y) < int(x):
            less += 1
    if less == mid:
        print(x)
```
