```python
class Box:

    def __init__(self, cat=None):
        self.cat = cat
        self.nextcat = None
        self.prevcat = None


class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None
        self._size = 0

    def __len__(self):
        return self._size

    def _check_index(self, index, allow_end=False):
        limit = self._size if allow_end else self._size - 1
        if not isinstance(index, int):
            raise TypeError("Индекс должен быть целым числом")
        if index < 0 or index > limit:
            raise IndexError(f"Индекс {index} вне диапазона")

    def _box_at(self, index):
        if index <= self._size // 2:
            box = self.head
            for _ in range(index):
                box = box.nextcat
        else:
            box = self.tail
            for _ in range(self._size - 1 - index):
                box = box.prevcat
        return box

    def insert(self, index, cat):
        self._check_index(index, allow_end=True)
        newbox = Box(cat)

        if self._size == 0:
            self.head = self.tail = newbox
        elif index == 0:
            newbox.nextcat = self.head
            self.head.prevcat = newbox
            self.head = newbox
        elif index == self._size:
            newbox.prevcat = self.tail
            self.tail.nextcat = newbox
            self.tail = newbox
        else:
            nextbox = self._box_at(index)
            prevbox = nextbox.prevcat
            newbox.prevcat = prevbox
            newbox.nextcat = nextbox
            prevbox.nextcat = newbox
            nextbox.prevcat = newbox

        self._size += 1

    def addToEnd(self, cat):
        self.insert(self._size, cat)

    def get(self, index):
        self._check_index(index)
        return self._box_at(index).cat

    def __getitem__(self, index):
        return self.get(index)

    def remove(self, index):
        self._check_index(index)
        box = self._box_at(index)

        if box.prevcat is not None:
            box.prevcat.nextcat = box.nextcat
        else:
            self.head = box.nextcat

        if box.nextcat is not None:
            box.nextcat.prevcat = box.prevcat
        else:
            self.tail = box.prevcat

        self._size -= 1
        return box.cat

    def __delitem__(self, index):
        self.remove(index)

    def contains(self, cat):
        box = self.head
        while box:
            if box.cat == cat:
                return True
            box = box.nextcat
        return False

    __contains__ = contains

    def __iter__(self):
        box = self.head
        while box:
            yield box.cat
            box = box.nextcat

    def __repr__(self):
        return "DoublyLinkedList([" + ", ".join(repr(c) for c in self) + "])"


if __name__ == "__main__":
    pass
```
