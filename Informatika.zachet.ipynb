```python
print("======================================")
print("   ТЕСТ ПО ИНФОРМАЦИОННОЙ БЕЗОПАСНОСТИ")
print("======================================")
print("В тесте 5 вопросов.")
print("На каждый вопрос дается 5 попыток.")
print("За каждый правильный ответ начисляется 1 балл.\n")

score = 0
max_score = 5
attempts_max = 5

questions = [
    {
        "question": "1. Что такое информационная безопасность?",
        "options": [
            "1 - Защита информации от несанкционированного доступа",
            "2 - Увеличение скорости работы компьютера"
        ],
        "answer": "1"
    },
    {
        "question": "2. Какой пароль является наиболее надежным?",
        "options": [
            "1 - 123456",
            "2 - R7#mP2!kL9"
        ],
        "answer": "2"
    },
    {
        "question": "3. Что следует сделать при получении письма с подозрительной ссылкой?",
        "options": [
            "1 - Сразу перейти по ссылке",
            "2 - Не переходить по ссылке и проверить отправителя"
        ],
        "answer": "2"
    },
    {
        "question": "4. Для чего нужна антивирусная программа?",
        "options": [
            "1 - Для обнаружения и удаления вредоносных программ",
            "2 - Для увеличения размера экрана"
        ],
        "answer": "1"
    },
    {
        "question": "5. В каком году родился Жорес Алфёров, лауреат Нобелевской премии по физике?",
        "options": [
            "Подсказка: он родился в первой половине XX века.",
            "Введите год рождения числом."
        ],
        "answer": "1930"
    }
]

for item in questions:
    print("\n" + item["question"])

    for option in item["options"]:
        print(option)

    attempts = attempts_max

    while attempts > 0:
        user_answer = input("Ваш ответ: ").strip()

        if user_answer == item["answer"]:
            print("Верно! Молодец!")
            score += 1
            break
        else:
            attempts -= 1

            if attempts > 0:
                print("Неверно, попробуйте снова!")
                print("Осталось попыток:", attempts)
            else:
                print("Попытки закончились.")
                print("Правильный ответ:", item["answer"])

print("\n======================================")
print("             ИТОГИ ТЕСТА")
print("======================================")

print("Набрано баллов:", score, "из", max_score)

if score == 5:
    grade = 5
elif score == 4:
    grade = 4
elif score == 3:
    grade = 3
else:
    grade = 2

print("Ваша оценка:", grade)

if grade == 5:
    print("Отличный результат!")
elif grade == 4:
    print("Хороший результат!")
elif grade == 3:
    print("Удовлетворительный результат.")
else:
    print("Рекомендуется повторить материал.")

print("Тестирование завершено!")
```
