1. Рекурсия
Стек — структура данных LIFO (Last-In First-Out). Стек вызовов состоит из стековых кадров; при каждом вызове функции создаётся новый кадр.

Рекурсия — приём, когда функция вызывает саму себя. Обязательны:

базовый (предельный) случай — условие выхода;

рекурсивный случай — вызов с изменёнными аргументами, приближающий к базовому.

Глубина рекурсии — максимальное число одновременно вложенных вызовов (размер стека).
Прямой ход — код до рекурсивного вызова; обратный ход — код после возврата из рекурсии.

```python
# Сумма 1..x
def f(x):
    if x == 1:            # базовый случай
        return 1
    return x + f(x - 1)   # рекурсивный случай

# Факториал с показом прямого/обратного хода
def fact(x):
    if x == 0:
        return 1
    print('Прямой ход:', x)
    res = x * fact(x - 1)
    print('Обратный ход:', x)
    return res

# Числа Фибоначчи (наивно — экспоненциально!)
def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib(n - 1) + fib(n - 2)

# Быстрое возведение в степень: O(log n)
def bin_pow(a, n):
    if n == 0:
        return 1
    if n % 2:                       # нечётная степень
        return bin_pow(a, n - 1) * a
    else:                           # чётная
        return bin_pow(a, n // 2) ** 2

print(bin_pow(2, 100))  # 1267650600228229401496703205376
```

```python
def bin_search(a, x):
    left, right = 0, len(a) - 1
    while left <= right:
        mid = (left + right) // 2
        if a[mid] == x:
            return mid
        elif a[mid] < x:
            left = mid + 1
        else:
            right = mid - 1
    return -1

a = [2, 5, 8, 12, 16, 23, 38, 56, 72]
print(bin_search(a, 23))   # 5
print(bin_search(a, 10))   # -1
```

```python
# Мемоизация через кэш-массив
def fib(n, cache):
    if cache[n] is not None:
        return cache[n]
    if n == 0:
        cache[n] = 0
        return 0
    elif n == 1:
        cache[n] = 1
        return 1
    cache[n] = fib(n - 1, cache) + fib(n - 2, cache)
    return cache[n]

n = 40
cache = [None] * (n + 1)
print(fib(n, cache))   # 102334155

# Итеративный вариант (динамика)
N = 40
fib_arr = [None] * (N + 1)
fib_arr[0], fib_arr[1] = 0, 1
for i in range(2, N + 1):
    fib_arr[i] = fib_arr[i - 2] + fib_arr[i - 1]
print(fib_arr[40])     # 102334155
```

Кузнечик: количество способов

```python
n = int(input())
a = [None] * (n + 1)
a[0] = 1
for i in range(1, n + 1):
    if i == 1:
        a[i] = a[0]
    else:
        a[i] = a[i - 1] + a[i - 2]

print(a[n])
```

Кузнечик со штрафом (минимальная стоимость пути)

```python
n = int(input())                        # число ступенек
c = list(map(int, input().split()))     # стоимости
a = [None] * (n + 1)
a[0] = c[0]
for i in range(1, n + 1):
    if i == 1:
        a[i] = a[i - 1] + c[i]
    elif i == 2:
        a[i] = min(a[i - 1], a[i - 2]) + c[i]
    else:
        a[i] = min(a[i - 1], a[i - 2], a[i - 3]) + c[i]

print(a[n - 1])
```

Восстановление пути

```python
n = int(input())
c = list(map(int, input().split()))
a, path = [None] * n, [None] * n
a[0], path[0] = c[0], -1
if n > 1:
    a[1], path[1] = a[0] + c[1], 0
    for i in range(2, n):
        if a[i - 1] < a[i - 2]:
            a[i], path[i] = a[i - 1] + c[i], i - 1
        else:
            a[i], path[i] = a[i - 2] + c[i], i - 2

current, result_path = n - 1, []
while current >= 0:
    result_path.append(current)
    current = path[current]
result_path.reverse()
print("Min cost:", a[n - 1])
print("Path:", result_path)
```

```python
class Box:
    def __init__(self, cat=None):
        self.cat = cat
        self.nextcat = None

class LinkedList:
    def __init__(self):
        self.head = None

    def addToEnd(self, newcat):
        newbox = Box(newcat)
        if self.head is None:
            self.head = newbox
            return
        last = self.head
        while last.nextcat:
            last = last.nextcat
        last.nextcat = newbox

    def contains(self, cat):
        cur = self.head
        while cur:
            if cur.cat == cat:
                return True
            cur = cur.nextcat
        return False

    def removeBox(self, rm_cat):
        cur = self.head
        if cur is not None:
            if cur.cat == rm_cat:
                self.head = cur.nextcat
                return
        prev = None
        while cur:
            if cur.cat == rm_cat:
                break
            prev = cur
            cur = cur.nextcat
        if cur is None:
            return
        prev.nextcat = cur.nextcat
```
